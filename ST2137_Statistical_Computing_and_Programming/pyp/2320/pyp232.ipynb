```python
import pandas as pd
import numpy as np

liverpool = pd.read_csv("liverpool_2223_season.csv")
goal_counts =pd.DataFrame(np.zeros((10, 2), dtype='int'),
columns=['GF', 'GA'])
tmp2 =liverpool.GF.value_counts()
goal_counts.loc[tmp2.index, 'GF'] =tmp2

tmp3 =liverpool.GA.value_counts()
goal_counts.loc[tmp2.index, 'GA'] =tmp3
goal_counts.plot(kind="bar", rot=1, title="Liverpool Goal Counts")
```
