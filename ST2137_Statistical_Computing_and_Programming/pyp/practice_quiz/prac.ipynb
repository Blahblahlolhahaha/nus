```python
import pandas as pd
import numpy as np
```

```python
def weighted_jacc(list1, list2):
    # Obtain counts from each set:
    A = pd.Series(list1).value_counts()
    B = pd.Series(list2).value_counts()
    
    df1 = pd.DataFrame({'x':A, 'y':B})
    
    # Replace missing values in df1 with 0:
    sad = df1.fillna(0)
    min_col = sad.min(axis=1)
    max_col = sad.max(axis=1)
    # Compute and return numerator/denominator:
    return np.sum(min_col) / np.sum(max_col)
```

```python


l1 = ['apple', 'apple', 'orange', 'durian', 'durian', 'mango']
l2 = ['apple', 'orange', 'orange', 'durian']

print(weighted_jacc(l1, l2))
print(3/7)
```

```python
from scipy import stats

student = pd.read_csv("student-mat.csv", sep=";")
table = pd.crosstab(student.address, student.guardian)
chisq_output = stats.chi2_contingency(table)
print(f"The p-value is {chisq_output.pvalue:.3f}.")
print(f"The test-statistic value is {chisq_output.statistic:.3f}.")
chisq_output.expected_freq
```

```python
def recurse(x):
    if x == 1:
        return 0
    if x == 2:
        return 1
    return 2 * recurse(x-1) - recurse(x-2)

res = []
for i in range(1, 31):
    res.append(recurse(i))
print(sum(res))
```
