```python
import pandas as pd
import numpy as np
```

```python
liverpool = pd.read_csv("liverpool_2223_season.csv")
goal_counts =pd.DataFrame(np.zeros((10, 2), dtype='int'),
columns=['GF', 'GA'])
tmp2 = liverpool.GF.value_counts()
goal_counts.loc[tmp2.index, 'GF'] =tmp2

tmp3 = liverpool.GA.value_counts()
goal_counts.loc[tmp2.index, 'GA'] =tmp3
goal_counts.plot(kind="bar",title="Liverpool Goal Counts")
```

```python
def estimate_lambda(y, tol=1e-6):
    lamb_old = y.mean()
    lamb_new = np.inf
    diff = lamb_new - lamb_old
    while diff > tol:
        lamb_new = y.mean() * (1-np.exp(-lamb_old))
        diff = np.abs(lamb_new - lamb_old)
        lamb_old = lamb_new
    return lamb_new
```

```python
y = np.repeat([1,2,3,5.0], [12, 14 ,3 ,1])

print(estimate_lambda(y))
```
