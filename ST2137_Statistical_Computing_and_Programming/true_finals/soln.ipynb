```python
def calc_yi(y):
    if y > 5:
        return (np.exp(-0.5 * (y ** 2)) / np.sqrt(2 * np.pi)) / norm.pdf(y, loc=5)
    return 0

rng = np.random.default_rng(2137)

p_mt5 = 1 - norm.cdf(5)
print(f"P(Z > 5) = {p_mt5}")

first_test = norm.rvs(random_state=rng, size=(1000,))
print(f"Point estimate of P(Z > 5) = {len(first_test[first_test > 5] / len(first_test))}")

second_test = norm.rvs(random_state=rng, loc=5, size=(1000,))
second_test_hi = list(map(calc_yi, second_test))
p_mt5_simul = 1/1000 * np.sum(second_test_hi)
print(f"P(Z > 5) (simulated) = {p_mt5}")
```
