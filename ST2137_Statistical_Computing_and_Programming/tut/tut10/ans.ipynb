```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
crabs = pd.read_csv("crab.txt", sep=" ")
crabs["spine"] = crabs["spine"].astype("category")
model = smf.ols("weight ~ width * spine", data=crabs).fit()

print(model.summary())
```

```python
p_value = model.pvalues["Intercept"]
beta4 = model.params["width:spine[T.2]"]
resi = model.resid

sigma2 = model.scale
r2 = model.rsquared_adj
sigma2

model.predict({
    "width": 27,
    "spine": 1
})
```
