```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(2138)
h = 0.3
def func(x):
    return x * np.exp((h * x ** 2 / 2))

sample_means = []
sample_trimmed_mean = []

for _ in range(50):
    total = []
    total_np = func(rng.normal(size=30))
    sample_means.append(np.mean(total_np))
    sample_trimmed_mean.append(stats.trim_mean(total_np, 0.1))

print("Mean: " + str(np.mean(sample_means)))
print("SD: " + str( np.std(sample_means)))

print("Mean (trimmed): " + str(np.mean(sample_trimmed_mean)))
print("SD (trimmed): " + str(np.std(sample_trimmed_mean)))
```

```python
def out_means_variance(x, x_mean, x_std, K = 2.24):
    return np.abs(x - x_mean) / x_std > K

def out_iqr(x, lower, upper, iqr):
    return x < (lower - iqr * 1.5) or x > (upper + iqr * 1.5)

def out_mad(x, median, mad, K = 2.24):
    return np.abs(x - median) / (mad / 0.6745) > K
```

```python
data = np.array([2, 2, 3, 3, 3, 4, 4, 4, 100000, 100000])
mean = np.mean(data)
std = np.std(data, ddof = 1)
print(list(filter(lambda x: out_means_variance(x, mean, std), data)))

lower = np.quantile(data, 0.25)
upper = np.quantile(data, 0.75)
iqr = upper - lower
print(list(filter(lambda x: out_iqr(x, lower, upper, iqr), data)))

median = np.median(data)
data_mad = np.abs(data - median)
mad_median = np.median(data_mad)
print(list(filter(lambda x: out_mad(x, median, mad_median), data)))
```

```python
from scipy.stats.mstats import winsorize
import pandas as pd

stud = pd.read_csv("student-mat.csv",sep=";")

g3_grp = stud.groupby("Medu")["G3"]

for medu, group in g3_grp:
    sample_mean = group.mean()
    win_mean = np.mean(winsorize(group, limits = 0.1))
    trim_mean = stats.trim_mean(group, 0.1)
    print(f"Medu {medu}:\nmean:{sample_mean:.2f}\nWinsorize mean:{win_mean:.2f}\nTrimmed mean:{trim_mean:.2f}\n")
```
