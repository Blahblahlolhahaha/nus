```python
import numpy as np
from scipy.stats import expon

rng = np.random.default_rng(2137)

x_i = expon.rvs(scale=1/2, random_state=rng, size=(1000,))

y_i = np.sin(x_i)
y_hat = np.mean(y_i)
se = 1.96 * np.std(y_i, ddof=1) / np.sqrt(len(x_i))
lower = y_hat - se
upper = y_hat + se

print(f"CI: ({lower},{upper})")

x_1 = expon.rvs(scale=1/2, random_state=rng, size=(1,))
y_hat = np.sin(x_1)
s_j  = 1
first = True
j = 1
while first or s_j/j >= 0.01:
    if first:
        first = False
    x_i = expon.rvs(scale=1/2, random_state=rng, size=(1,))
    y_j = np.sin(x_i)
    y_hat_o = y_hat
    y_hat = y_hat + (y_j - y_hat) / (j + 1)
    s_j = (1 - 1 / j) * s_j + (j + 1) * (y_hat - y_hat_o) ** 2
    j += 1
print(j)
print(y_hat)
```
