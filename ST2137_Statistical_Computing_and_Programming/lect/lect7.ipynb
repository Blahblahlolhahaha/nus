```python
import pandas as pd
import numpy as np
from scipy import stats

abalone = pd.read_csv("data/abalone_sub.csv")
hr = pd.read_csv("data/health_promo_hr.csv")

x = abalone.viscera[abalone.gender == "M"]
y = abalone.viscera[abalone.gender == "F"]

t_out = stats.ttest_ind(x,y)
ci_95 = t_out.confidence_interval()
print(f"""
Pvalue for the test is {t_out.pvalue:.3f}
Test statistics {t_out.statistic:.3f}
95% CI: {tuple(list(ci_95))}
""")
```
