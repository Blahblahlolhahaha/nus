```python
import pandas as pd
import numpy as np
from scipy import stats

chem = pd.read_csv("data/mass_chem.csv")

chem.chem.mean()
```

```python
stats.trim_mean(chem.chem, 0.1)
```

```python
win = stats.mstats.winsorize(chem.chem, 0.1) # different from r due to different quantile used!
print(sorted(win))
print(win.mean()) 
```

```python
awareness = np.array([77, 87, 88, 114, 151, 210, 219, 246, 253, 262, 296
               , 299, 306, 376, 428, 515, 666, 1310, 2611])

print(awareness.std()) # python uses 1/n instead of 1/(n-1)

print(awareness.std(ddof=1)) # set denominator degrees of freedom
```

```python
stats.median_abs_deviation(awareness)
```

```python
stats.iqr(awareness)
```
